### Вправа 1. Рядок чека (f-рядки)

Напиши функцію: за назвою послуги й сумою збери рядок **ширини 28**: назва вирівняна ліворуч у 18 символах, сума — праворуч у 10 символах з двома знаками після коми.

`receipt_line("Доставка", 95)` → `"Доставка               95.00"`

```python
def receipt_line(service, amount):
    return f"{service:<18}{amount:>10.2f}"


print(repr(receipt_line("Доставка", 95)))
print('receipt_line("Доставка", 95) == "Доставка               95.00" →', receipt_line("Доставка", 95) == "Доставка               95.00")
print('receipt_line("Пакування", 12.5) == "Пакування              12.50" →', receipt_line("Пакування", 12.5) == "Пакування              12.50")
print('len(receipt_line("Страхування", 3.755)) == 28 →', len(receipt_line("Страхування", 3.755)) == 28)
```

### Вправа 2. Нормалізація телефону

Телефони приходять у різному вигляді. Приведи номер до формату `+380XXXXXXXXX`:

- прибери пробіли по краях, а також усі пробіли, дефіси й дужки всередині;
- якщо номер починається з `0` — додай `+38`;
- якщо з `380` — додай `+`.

```python
def normalize_phone(raw):
    raw = f"{raw.replace(' ', '').replace('-', '').replace(')', '').replace('(', '')}"
    if raw.startswith("0"):  raw = "+38" + raw
    if raw.startswith("380"):  raw = "+" + raw
    return raw

for raw in ["  050-123-45-67 ", "(067) 000 11 22", "380931112233", "+380501234567"]:
    print(f"{raw!r:22} → {normalize_phone(raw)}")
print('normalize_phone("  050-123-45-67 ") == "+380501234567" →', normalize_phone("  050-123-45-67 ") == "+380501234567")
print('normalize_phone("(067) 000 11 22") == "+380670001122" →', normalize_phone("(067) 000 11 22") == "+380670001122")
print('normalize_phone("380931112233") == "+380931112233" →', normalize_phone("380931112233") == "+380931112233")
print('normalize_phone("+380501234567") == "+380501234567" →', normalize_phone("+380501234567") == "+380501234567")
```

### Вправа 3. Зрізи: ініціали й маска картки

1. `initials("Олена Ковальчук")` → `"О. К."` (підказка: `split()` і `[0]`).
2. `mask_card("4441 1144 5566 1234")` → `"**** **** **** 1234"` — лишити тільки останні 4 символи.

```python
def initials(full_name):
    return f"{full_name.split()[0][0]}. {full_name.split()[1][0]}."



def mask_card(card):
    return f"{'**** ' * 3}{card.split()[3]}"


print(initials("Олена Ковальчук"), mask_card("4441 1144 5566 1234"))
print('initials("Олена Ковальчук") == "О. К." →', initials("Олена Ковальчук") == "О. К.")
print('initials("Тарас Шевченко") == "Т. Ш." →', initials("Тарас Шевченко") == "Т. Ш.")
print('mask_card("4441 1144 5566 1234") == "**** **** **** 1234" →', mask_card("4441 1144 5566 1234") == "**** **** **** 1234")
```

### Вправа 4. Вага з тексту

Ваги приходять рядками виду `"Вага: 12,5 кг"`. Витягни число як `float`: прибери підпис, одиницю, пробіли, заміни кому на крапку.

```python
def parse_weight(text):
    return float(text.split()[1].replace(',', '.'))


print(parse_weight("Вага: 12,5 кг"))
print('parse_weight("Вага: 12,5 кг") == 12.5 →', parse_weight("Вага: 12,5 кг") == 12.5)
print('parse_weight("Вага: 3 кг") == 3.0 →', parse_weight("Вага: 3 кг") == 3.0)
print('isinstance(parse_weight("Вага: 0,75 кг"), float) →', isinstance(parse_weight("Вага: 0,75 кг"), float))
```

### Вправа 5. Сума цифр трекінг-номера

Порахуй суму цифр **цілого** числа, використовуючи лише `%` і `//` (без перетворення на рядок). Підказка: остання цифра — `n % 10`, решта — `n // 10`; повтори для п'ятизначного числа п'ять разів.

```python
def digit_sum_5(n):
    """Сума цифр п'ятизначного числа n."""
    return n//10000 + n//1000%10 + n//100%10 + n//10%10 + n%10


print(digit_sum_5(20451))
print('digit_sum_5(20451) == 12 →', digit_sum_5(20451) == 12)
print('digit_sum_5(99999) == 45 →', digit_sum_5(99999) == 45)
print('digit_sum_5(10000) == 1 →', digit_sum_5(10000) == 1)
print('digit_sum_5(20451) == sum(int(c) for c in str(20451)) →', digit_sum_5(20451) == sum(int(c) for c in str(20451)))
```

### Вправа 6. Безпечна кількість

`parse_count(raw)` повертає `int`, якщо після `strip()` у рядку лише цифри, інакше — `0`. Без `try/except`.

```python
def parse_count(raw):
    if raw.strip().isdigit():
       return int(raw.strip())
    else:
        return 0


for raw in [" 12 ", "7\n", "-3", "2.5", "", "abc"]:
    print(f"{raw!r:8} → {parse_count(raw)}")
print('parse_count(" 12 ") == 12 and parse_count("7\\n") == 7 →', parse_count(" 12 ") == 12 and parse_count("7\n") == 7)
print('parse_count("-3") == 0 and parse_count("2.5") == 0 and parse_count("") == 0 →', parse_count("-3") == 0 and parse_count("2.5") == 0 and parse_count("") == 0)
print('isinstance(parse_count("5"), int) →', isinstance(parse_count("5"), int))
```

### Вправа 7. Порівняння сум

Кур'єр отримав три оплати `0.1`, `0.2` і `0.3` грн (тестові перекази). Напиши `same_amount(a, b)`, що правильно порівнює дві суми `float`, і `to_kop(amount)` — перетворює гривні в копійки (`int`) з округленням.

```python
import math


def same_amount(a, b):
    return round(a, 2) == round(b, 2)


def to_kop(amount):
    return int(round(amount * 100))


print(0.1 + 0.2, same_amount(0.1 + 0.2, 0.3), to_kop(0.1 + 0.2))
print('same_amount(0.1 + 0.2, 0.3) →', same_amount(0.1 + 0.2, 0.3))
print('not same_amount(0.3, 0.31) →', not same_amount(0.3, 0.31))
print('to_kop(0.1 + 0.2) == 30 and to_kop(12.5) == 1250 →', to_kop(0.1 + 0.2) == 30 and to_kop(12.5) == 1250)
print('isinstance(to_kop(1.99), int) and to_kop(1.99) == 199 →', isinstance(to_kop(1.99), int) and to_kop(1.99) == 199)
```

### Вправа 8. SMS у два рядки

Напиши `sms(name, office, minutes)`, що повертає **два рядки**, розділені `\n`:

1. `Привіт, <name>!`
2. `Відділення №<office>:` + **табуляція** + `чекаємо <minutes> хв`

`sms("Олена", 12, 30)` → `"Привіт, Олена!\nВідділення №12:\tчекаємо 30 хв"`

Збери рядок з кількох f-рядків — сусідніми шматками в дужках або через `"\n".join`.

```python
def sms(name, office, minutes):
    record = [f"Привіт, {name}!",f"Відділення №{office}:\tчекаємо {minutes} хв"]
    return "\n".join(record)


text = sms("Олена", 12, 30)
print(text)
print(repr(text))
print('text == "Привіт, Олена!\\nВідділення №12:\\tчекаємо 30 хв" →', text == "Привіт, Олена!\nВідділення №12:\tчекаємо 30 хв")
print('text.splitlines() == ["Привіт, Олена!", "Відділення №12:\\tчекаємо 30 хв"] →', text.splitlines() == ["Привіт, Олена!", "Відділення №12:\tчекаємо 30 хв"])
print('sms("Тарас", 3, 5).count("\\t") == 1 →', sms("Тарас", 3, 5).count("\t") == 1)
```
