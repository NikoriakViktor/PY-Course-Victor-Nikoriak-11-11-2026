## Як влаштований Jupyter Notebook

Цей файл, який ти зараз читаєш, — і є **notebook**: документ, що складається з окремих блоків — **cells** (клітинок).

Є два типи клітинок:

- **Markdown cell** — текст, як цей: пояснення, умови завдань, посилання. Не виконується, а лише форматується.
- **Code cell** — Python-код. Виконується окремо від інших клітинок.

Щоб виконати code cell:

1. Клікни всередину клітинки.
2. Натисни **Shift + Enter** (виконає клітинку і перейде до наступної) або **Ctrl + Enter** (виконає і залишиться на місці).
3. Результат (`Output`) з'явиться одразу під клітинкою.

Клітинку можна запускати повторно й скільки завгодно разів: зміниш код і запустиш ще раз — старий Output заміниться новим.

## AI-політика курсу

AI-асистент (Claude Code, Copilot, ChatGPT тощо) — це **співрозробник, а не заміна**. Курс визначає два режими роботи, і на кожному занятті буде зрозуміло, який саме зараз діє.

### Режим 1 — Самостійна перевірка (без AI)

Коротке завдання без AI: пояснити код, знайти помилку. Мета — перевірити, що ти сам розумієш, що відбувається, без підказки.

**Приклад:** на занятті викладач показує невеликий фрагмент коду з помилкою і просить усно пояснити, що станеться при запуску, — до того, як хтось його запустить.

### Режим 2 — Робота з AI (з перевіркою)

Можна отримувати підказки й варіанти рішення від AI. Але результат проходить перевірку, а не приймається "на віру". Робочий цикл:

```text
ASK → PREDICT → RUN → VERIFY → EXPLAIN → ADAPT
```

1. **Ask** — постав запитання чи попроси варіант рішення в AI.
2. **Predict** — перед тим як запускати, подумай, що має статися.
3. **Run** — запусти код.
4. **Verify** — перевір: результат збігається з прогнозом? Тести чи приклади проходять?
5. **Explain** — поясни своїми словами, що робить код рядок за рядком.
6. **Adapt** — зміни рішення під іншу вимогу (наприклад, викладач попросить змінити одну деталь — і ти маєш зрозуміти, що саме змінити).

Якщо студент доходить до кроку Adapt і може змінити рішення сам(-а) — це і є доказ, що він(-она) реально зрозумів(-ла) код, а не просто скопіював(-ла) його.

**Приклад:** ти просиш AI пояснити повідомлення про помилку — це нормально (Режим 2). Але вставити готовий код домашнього завдання без прогнозу результату — це саме той момент, де AI забирає в тебе навичку, яку урок мав розвинути.

#### Корисне й некорисне використання AI

| Ситуація | Корисне використання | Некорисне використання |
|---|---|---|
| Помилка (error) | «Поясни, що означає ця помилка» | «Зроби все завдання замість мене» |
| Новий код | «Поясни цей рядок» | Сліпо вставити 40 рядків без прогнозу |
| Домашня робота | Попросити натяк або альтернативний підхід | Здати AI-код, який не можеш пояснити |

### Примітка в домашніх роботах

У кожному файлі, який здаєш, додавай рядок-коментар:

```python
# AI log: одне речення — що саме питав у AI і що написав сам
```

Якщо AI не використовував(-ла) — так і пиши:

```python
# AI log: none
```

Це не контроль заради покарання, а прозорість і рефлексія: рядок фіксує, що саме зробив AI, а що — ти сам(-а), і привчає усвідомлено користуватися AI як інструментом, а не як заміною власного мислення.

Оцінюється і працездатність рішення, і здатність його пояснити — тому "готовий код без розуміння" не проходить перевірку навіть якщо він працює.

> Ця політика діє в **усіх** модулях курсу, не лише в бонусному AI-треку (`Self-paced. AI для розробників`) наприкінці програми.

## Git та GitHub — коротко

Два різні поняття, які легко сплутати:

```text
Git     = система керування історією змін (працює локально, на твоєму комп'ютері)
GitHub  = сервіс, де живе віддалена копія репозиторію (для співпраці й здачі робіт)
```

Ми працюємо через GitHub, як у реальній IT-команді:

```text
репозиторій курсу → fork → clone → локальна робота → гілка → commit → push → Pull Request
```

Pull Request (PR) = цільовий спосіб здачі домашньої роботи в цьому курсі.
