## 1. Що таке ноутбук

**Ноутбук** (Jupyter Notebook, файл `.ipynb`) — це документ, складений з **клітинок**. Клітинки бувають двох видів:

| Клітинка | Що в ній | Що відбувається після запуску |
|---|---|---|
| **Code** | код Python | Python виконує код, результат з'являється під клітинкою |
| **Markdown** (у Colab — **Text**) | текст із форматуванням, як цей | текст гарно відображається |

Ти зараз читаєш **markdown-клітинку**. Двічі клацни по ній — побачиш, як вона написана «зсередини». Щоб повернути гарний вигляд, натисни **Shift+Enter**.

## 2. Що таке Google Colab

**Google Colab** (Colaboratory) — безкоштовний сервіс Google, який запускає ноутбуки **просто в браузері**:

- нічого не треба встановлювати — Python уже готовий;
- потрібен лише Google-акаунт;
- код виконується не на твоєму комп'ютері, а на тимчасовій машині Google у хмарі (її називають **runtime**, середовище виконання).

### Найважливіше в Colab

| Що зробити | Як |
|---|---|
| **зберегти свою копію** — зроби це першим! | **File → Save a copy in Drive** |
| виконати клітинку | **Shift+Enter** або кнопка ▶ ліворуч від клітинки |
| додати клітинку з кодом | кнопка **+ Code** угорі |
| додати текстову клітинку | кнопка **+ Text** угорі |
| виконати всі клітинки згори донизу | **Runtime → Run all** |
| почати «з чистого аркуша» | **Runtime → Restart session** |
| зупинити код, що «завис» | **Runtime → Interrupt execution** |

> ⚠️ **Важливо.** Ноутбук, відкритий за посиланням із GitHub, **не зберігає твоїх змін**. Спершу **File → Save a copy in Drive** — копія з'явиться на твоєму Google Drive у папці «Colab Notebooks», і далі працюй у ній.

⏳ Якщо ти довго нічого не робиш, Colab відключає машину. Твій код у ноутбуці залишиться, але змінні зникнуть — просто запусти клітинки знову (**Runtime → Run all**).

````python
print("Hello, курс Python у Beetroot Academy!")
````

## 5. Змінні: ноутбук пам'ятає

**Змінна** — це ім'я, до якого прив'язане значення. Запусти клітинку нижче:

````python
course = "Python"
school = "Beetroot Academy"
````

Під клітинкою нічого не з'явилося — і це нормально: ми лише запам'ятали значення. А тепер використаємо їх **в іншій клітинці**:

````python
print("Я вивчаю", course, "у", school)
````

Змінні, створені в одній клітинці, доступні в усіх інших — **але тільки після того, як ту клітинку виконано**.

**Експеримент:** у Colab натисни **Runtime → Restart session**, а потім одразу запусти клітинку з `print("Я вивчаю", ...)`. Побачиш помилку `NameError: name 'course' is not defined`: після перезапуску пам'ять порожня. Виправлення — запусти клітинки по порядку згори донизу.

Ще один експеримент. Запусти клітинку нижче **тричі поспіль** і подивись на число:

````python
counter = 0
````

````python
counter = counter + 1
print("Клітинку запущено разів:", counter)
````

Число росте, бо ноутбук пам'ятає `counter` між запусками. Важливий не порядок клітинок на сторінці, а **порядок, у якому ти їх запускав**. Номер у квадратних дужках ліворуч від клітинки `[5]` показує, якою за рахунком вона виконалась.

## 6. Markdown: як писати текст

Двічі клацни по цій клітинці, щоб побачити розмітку:

- `# Заголовок`, `## Підзаголовок`
- **жирний** — `**жирний**`, *курсив* — `*курсив*`
- `код` — у зворотних апострофах
- списки — рядки, що починаються з `-` або `1.`
- посилання — `[текст](https://beetroot.academy)`

**Спробуй:** додай нижче клітинку **+ Text** і напиши в ній 2–3 рядки про себе, використовуючи заголовок і список.
